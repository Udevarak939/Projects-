# Import Libraries

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

# Load The Data


```python
df = pd.read_csv('/content/amazon_sales_data 2025.csv')
```

# Convert date column to datetime format

```python
df['Date'] = pd.to_datetime(df['Date'])
```

# Basic dataset info

```python
df.info()
```

# Check for missing values

```python
df.isnull().sum()
```

# Summary statistics

```python
df.describe()
```

#Sales Trends Over Time

```python
plt.figure(figsize=(10,5))
df.groupby('Date')['Total Sales'].sum().plot(kind='line', marker='o', color='b')
plt.title('Total Sales Over Time')
plt.xlabel('Date')
plt.ylabel('Sales Amount')
plt.grid(True)
plt.show()
```

Order Status Analysis: Completed vs Canceled


```python
# Distribution of Order Status
plt.figure(figsize=(7,5))
sns.countplot(x='Status', data=df, palette='Set2')
plt.title('Order Status Distribution')
plt.xlabel('Status')
plt.ylabel('Count')
plt.show()
```

```python
#Most Sold Products
plt. figure(figsize=(10,5))
top_products = df['Product'].value_counts().nlargest(10)
sns.barplot(x=top_products.index, y=top_products.values, palette='coolwarm')
plt.title('Top 10 Most Sold Products')
plt.xlabel('Product Name')
plt.ylabel('Number of Orders')
plt.xticks(rotation=35)
plt.show()
```

Total Revenue by Product Category

```python
#Total Revenue by Product Category
plt.figure(figsize=(10, 5))
category_sales = df.groupby('Category')['Total Sales'].sum().sort_values(ascending=False)
sns.barplot(x=category_sales.index, y=category_sales.values, palette='viridis')
plt.title('Total Sales by Category')
plt.xlabel('Category')
plt.ylabel('Total Sales')
plt.xticks(rotation=45)
plt.show()
```

Payment Method Distribution

```python
#Payment Method Distribution
plt.figure(figsize=(8, 5))
sns.countplot(y='Payment Method', data=df, palette='pastel')
plt.title('Payment Method Distribution')
plt.xlabel('Count')
plt.ylabel('Payment Method')
plt.show()
```

Sales by Customer Location

```python
#Sales by Customer Location
plt.figure(figsize=(10, 5))
top_locations = df.groupby('Customer Location')['Total Sales'].sum().nlargest(10)
sns.barplot(x=top_locations.values, y=top_locations.index, palette='magma')
plt.title('Top 10 Cities by Sales')
plt.xlabel('Total Sales')
plt.ylabel('City')
plt.show()
```

Sales Distribution by Order Quantity

```python
#Sales Distribution by Order Quantity
plt.figure(figsize=(7, 5))
sns.histplot(df['Quantity'], bins=10, kde=True, color='g')
plt.title('Order Quantity Distribution')
plt.xlabel('Quantity')
plt.ylabel('Frequency')
plt.grid(True)
plt.show()
```

Average order value

```python
#Average order value
avg_order_value = df['Total Sales'].mean()
print(f"\nAverage Order Value: ${avg_order_value:.2f}")
```

Most Popular Payment Method

```python
#Most Popular Payment Method
popular_payment = df['Payment Method'].value_counts().idxmax()
print(f"\nMost Used Payment Method: {popular_payment}")
```

Percentage of Completed Orders

```python
#Percentage of Completed Orders
completed_percentage = (df['Status'].value_counts(normalize=True)['Completed']) * 100
print(f"\nPercentage of Completed Orders: {completed_percentage:.2f}%")
```
